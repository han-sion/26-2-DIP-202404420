```python
f = plt.imread(imgpath+'/pollen-lowcontrast.tif')
print(f.shape,f.dtype)
```

```python
plt.imshow(f, cmap = 'gray', vmin=0, vmax=255)
plt.axis('off')
plt.colorbar()
plt.show()
```

```python
plt.hist(f.flatten())
plt.show()
```

```python
def histogram_equalization(image):
    histogram, bins = np.histogram(image.flatten(), bins=256, range=[0,256])
    cdf = histogram.cumsum()
    cdf_min = cdf.min()
    cdf_max = cdf.max()
    cdf_normalized = (cdf - cdf_min) * 255 / (cdf_max - cdf_min)
    cdf_normalized = cdf_normalized.astype(np.uint8)
    equalized_image = cdf_normalized[image]
    return equalized_image
```

```python
f_eq = histogram_equalization(f)
plt.hist(f_eq.flatten())
plt.show()
```

```python
plt.imshow(f_eq, cmap = 'gray', vmin=0, vmax=255)
plt.axis('off')
plt.colorbar()
plt.show()
```

### Exercise 1.
- 아래의 예는 local histogram equalization을 수행한다.
- 적절한 neighborhood size N을 찾아서 숨겨진 기호들을 모두 확인해보자.
- plt.imshow의 vim, vmax를 수정하여서도 숨겨진 기호들을 확인할 수 있으나, 본 실습에서는 vmax=255로 고정하여 histogram processing을 통해 숨겨진 기호들을 확인하자.

```python
N = 16
f_eq_local = np.zeros_like(f)
for x in np.arange(0,256,N):
  for y in np.arange(0,256,N):
    f_eq_local[x:x+N,y:y+N] = histogram_equalization(f[x:x+N,y:y+N])

plt.imshow(f_eq_local,vmin=0,vmax=255,cmap='gray')
plt.axis('off')
plt.show()

# 초기 N 값이 256 이고 이미지 크기가 256 * 256 이기 때문에 새 N을 256의 약수로 잡으면 모든 블록이
# 같은 크기로 딱 나누어 떨어져 가장자리에 크기가 다른 블록이 생기지않을 것 같다고 판단했다.
# N=256(영상 전체를 한 번에 처리)에서 시작해 128, 64, 32, 16, 8 순으로 줄여가며 결과를 비교했다.
# N이 클 때(128, 64, 32)는 숨겨진 기호가 잘 드러나지 않았다.
# N이 작을 때(8)는 화면이 지저분해져 기호를 알아보기 어려웠다.
# N=16일 때 숨겨진 기호들이 가장 잘 보여서 N=16으로 정했다.
```

# Linear Spatial Filtering
## 2D convolution

```python
import scipy.signal as signal
from skimage import io
```

```python
f = io.imread('https://press.hufs.ac.kr/data/file/gift/6f5a7159a604fe7de30498fd76c8741d_JIWi6Gj9_d6e43320ba8ee4ac452deadd5246a2d6db29e853.jpg',as_gray=True)[::2,::2]
print(f.shape,f.dtype)
```

```python
plt.imshow(f,cmap='gray')
plt.axis('off')
plt.show()
```

```python
w = np.random.randn(3,3)
#w = np.array([[-1,0,1],
#              [-2,0,2],
#              [-1,0,1]])
print(w)
```

```python
sz0,sz1 = f.shape
y = np.zeros_like(f)
for i in np.arange(1,sz0-1):
  for j in np.arange(1,sz1-1):
    y[i,j] = np.sum(f[i-1:i+2,j-1:j+2]*w[::-1,::-1])

plt.figure(dpi=150)
plt.subplot(1,2,1)
plt.imshow(f,cmap='gray')
plt.axis('off')
plt.title('original')
plt.subplot(1,2,2)
plt.imshow(y,cmap='gray')
plt.axis('off')
plt.title('filtered')
plt.show()
```

```python
y = signal.convolve2d(f,w,mode='same')
y.shape
plt.figure(dpi=150)
plt.subplot(121)
plt.imshow(f,cmap='gray')
plt.axis('off')
plt.title('original')
plt.subplot(122)
plt.imshow(y,cmap='gray')
plt.axis('off')
plt.title('filtered')
plt.show()
```

## lowpass filtering - Gaussian kernel

```python
N = 3
std = 2
s = np.arange(-N,N+1,1)
t = np.arange(-N,N+1,1)
S,T = np.meshgrid(s,t)
wg = np.exp(-(S**2+T**2)/(2*std*std))
wg = wg/np.sum(wg)

plt.figure(figsize=(7,7))
plt.imshow(wg,vmin=0,vmax=0.1,cmap='gray')
plt.axis('off')
plt.colorbar()
plt.show()
```

```python
y_g = signal.convolve2d(f,wg,mode='same')

plt.figure(figsize=(16,9))
plt.subplot(1,3,1)
plt.imshow(f,cmap='gray')
plt.axis('off')
plt.title('original')
plt.subplot(1,3,2)
plt.imshow(y_g,cmap='gray')
plt.axis('off')
plt.title('filtered')
plt.subplot(1,3,3)
plt.imshow(y_g-f,cmap='gray',vmin=-.01,vmax=.01)
plt.axis('off')
plt.title('difference')
plt.show()
```

```python
s = np.arange(-13,14,1)
t = np.arange(-13,14,1)
S,T = np.meshgrid(s,t)

for std in np.array([1,3,5,7]):
  wg = np.exp(-(S**2+T**2)/(2*std*std))
  wg = wg/np.sum(wg)

  y_g = signal.convolve2d(f,wg,mode='same')

  plt.figure(figsize=(16,9))
  plt.subplot(1,3,1)
  plt.imshow(f,cmap='gray')
  plt.axis('off')
  plt.title('original')
  plt.subplot(1,3,2)
  plt.imshow(y_g,cmap='gray')
  plt.axis('off')
  plt.title(f'filtered - sigma:{std}')
  plt.subplot(1,3,3)
  plt.imshow(y_g-f,cmap='gray',vmin=-.01, vmax=.01)
  plt.axis('off')
  plt.title('difference')
  plt.show()
```

```python
f = io.imread('/content/red_panda.png',as_gray=True)
print(f.shape,f.dtype)
```

```python
import time
```

```python
# 연산 시간: N이 커질수록 계산 시간이 눈에 띄게 늘어났다.
# 필터 크기가 N×N이라 한 픽셀을 계산할 때 더하는 값의 개수가 N²만큼 늘어나기 때문이라고 생각했다.
# (3×3은 9개, 30×30은 900개)
# filtered : N이 커질수록 사진이 점점 더 뿌옇게 흐려졌고, 작은 디테일과 선명한 경계가 사라졌다.
# 한 픽셀을 더 넓은 주변 영역의 평균으로 바꾸기 때문이라고 판단했다.
# difference : N이 커질수록 물체의 윤곽선이 더 굵고 진하게 나타났다.
# 평평한 부분은 흐려도 거의 변하지 않고, 밝기가 급하게 바뀌는 경계에서 차이가 크게 생기는 것으로 보인다.
# 가장자리: N이 커질수록 사진 테두리가 어두워지는 범위도 넓어졌다.
# 영상 바깥을 경계를 0으로 채우는 zero padding 방식이라서,
# 가장자리 픽셀의 평균에 0이 섞여 들어가기 때문이라고 판단했다. 커널이 클수록 0이 섞이는 범위도 넓어진다.

for N in np.array([3,7,15,30]):
  wb = np.ones((N,N))
  wb = wb/np.sum(wb)
  s0 = time.time()
  y_b = signal.convolve2d(f,wb,mode='same')
  compu_time=time.time() - s0
  print('computation time = ', compu_time,' sec')
  plt.figure(figsize=(16,9))
  plt.subplot(1,3,1)
  plt.imshow(f,cmap='gray')
  plt.axis('off')
  plt.title('original')
  plt.subplot(1,3,2)
  plt.imshow(y_b,cmap='gray')
  plt.axis('off')
  plt.title(f'filtered - box size: {N}')
  plt.subplot(1,3,3)
  plt.imshow(y_b-f,cmap='gray',vmin=-.01,vmax=.01)
  plt.axis('off')
  plt.title('difference')
  plt.show()
```

### Exercise 3.
- highpass filtering의 결과를 활용하여 blurred image를 sharp하게 개선시켜 보자.
- 자신만의 blurred image를 가져와서, Laplacian kernel을 통한 highpass filtering을 진행하고 이를 이용하여 원본영상의 sharpness를 증대시켜보자.
- w 및 c를 적절하게 변화시켜 가면서 결과를 확인하자.

```python
# 사진이 어둡고 해상도가 낮아 원래 노이즈가 많은 편이라,
# 경계를 너무 강하게 잡으면 노이즈까지 같이 커질 것 같다고 판단했다.
# 그래서 w는 대각선까지 포함하는 8방향 Laplacian kernel 대신,
# 효과가 상대적으로 완만한 4방향 Laplacian kernel을 사용했다.
# c는 값을 바꿔가며 결과를 비교했다.
# c가 작을 때는 원본과 거의 차이가 없었다.
# c가 클 때(예: 30)는 경계뿐 아니라 노이즈까지 과하게 강조되어 사진이 오히려 지저분해졌다.
# c=1일 때 고양이의 윤곽과 풀숲이 또렷해지면서도 노이즈가 크게 거슬리지 않아 c=1로 정했다.
# 원본과 결과를 같은 기준으로 비교하기 위해 두 영상 모두
# vmin=0, vmax=f.max()로 표시 범위를 맞췄다.


f = io.imread('/content/blurred_picture.jpg',as_gray=True)[::2,::2]

w = np.array([[0,-1,0],[-1,4,-1],[0,-1,0]])
y = signal.convolve2d(f,w,mode='same')
c = 1
g = f + c*(y)
plt.figure(figsize=(18,6))
plt.subplot(131)
plt.imshow(f, cmap='gray', vmin=0, vmax=f.max())   # original
plt.colorbar()
plt.axis('off')
plt.title('original')
plt.subplot(132)
plt.imshow(g, cmap='gray', vmin=0, vmax=f.max())   # sharpening
plt.colorbar()
plt.axis('off')
plt.title('sharpening')
plt.subplot(133)
plt.imshow(y,cmap='RdBu_r',vmin=-0.5*np.max(np.abs(y)),vmax=0.5*np.max(np.abs(y)))
plt.colorbar()
plt.axis('off')
plt.title('filtered')
plt.show()
```

## Sobel (1st order diff.)

```python
sobel1 = np.array([[-1,-2,-1],[0,0,0],[1,2,1]])
sobel2 = np.array([[-1,0,1],[-2,0,2],[-1,0,1]])
print(sobel1)
print(sobel2)
```

```python
g1 = signal.convolve2d(f,sobel1,mode='same')
g2 = signal.convolve2d(f,sobel2,mode='same')
G = np.sqrt(g1**2+g2**2)

plt.figure(figsize=(18,6))
plt.subplot(141)
plt.imshow(f,cmap='gray')
plt.colorbar()
plt.axis('off')
plt.title('original')
plt.subplot(142)
plt.imshow(g1,cmap='RdBu_r',vmin=-128,vmax=128)
plt.colorbar()
plt.axis('off')
plt.title('Gy')
plt.subplot(143)
plt.imshow(g2,cmap='RdBu_r',vmin=-128,vmax=128)
plt.colorbar()
plt.axis('off')
plt.title('Gx')
plt.subplot(144)
plt.imshow(G,cmap='gray',vmin=0,vmax=255)
plt.colorbar()
plt.axis('off')
plt.title('magnitude of gradient')
plt.show()
```

```python
f = plt.imread(os.path.join(imgpath,'testpattern512.tif'))

g1 = signal.convolve2d(f,sobel1,mode='same')
g2 = signal.convolve2d(f,sobel2,mode='same')
G = np.sqrt(g1**2+g2**2)

plt.figure(figsize=(18,6))
plt.subplot(131)
plt.imshow(f,cmap='gray')
plt.colorbar()
plt.axis('off')
plt.title('original')
plt.subplot(132)
plt.imshow(g1,cmap='RdBu_r',vmin=-128,vmax=128)
plt.colorbar()
plt.axis('off')
plt.title('Gy')
plt.subplot(133)
plt.imshow(g2,cmap='RdBu_r',vmin=-128,vmax=128)
plt.colorbar()
plt.axis('off')
plt.title('Gx')
plt.show()
```
